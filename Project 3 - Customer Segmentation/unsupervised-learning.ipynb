```python
cluster_df = df.drop(
    columns=[
        "OrderID",
        "CustomerID",
        "TrackingNumber",
        "ShippingAddress",
        "Date"
    ]
)

cluster_df.head()
```

```python
cluster_df.info()
```

```python
cluster_df = pd.get_dummies(
    cluster_df,
    columns=[
        "Product",
        "PaymentMethod",
        "OrderStatus",
        "CouponCode",
        "ReferralSource"
    ],
    drop_first=True
)

cluster_df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(cluster_df)
scaled_data.shape
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=0.95)
pca_data = pca.fit_transform(scaled_data)
pca_data.shape
```

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(pca_data)
    
    wcss.append(kmeans.inertia_)
    
import warnings
warnings.filterwarnings("ignore")  
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.title("Elbow Method")

plt.xlabel("Number of Clusters (K)")

plt.ylabel("WCSS")

plt.show()
```

```python
print(wcss)
```

```python
from sklearn.metrics import silhouette_score
silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(pca_data)

    score = silhouette_score(pca_data, labels)

    silhouette_scores.append(score)
```

```python
for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k}: {score:.4f}")
```

```python
kmeans = KMeans(
    n_clusters=6,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(pca_data)
cluster_df["Cluster"] = clusters
cluster_df.head()
```

```python
cluster_df["Cluster"].value_counts().sort_index()
```

```python
cluster_df.groupby("Cluster")[["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]].mean().round(2)
```

```python
from sklearn.decomposition import PCA
pca_vis = PCA(n_components=2)

pca_2d = pca_vis.fit_transform(scaled_data)
import pandas as pd

pca_df = pd.DataFrame(
    pca_2d,
    columns=["PC1", "PC2"]
)

pca_df["Cluster"] = clusters
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Cluster",
    palette="Set2",
    s=70
)

plt.title("Customer Segmentation using PCA and K-Means")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```
